### Step 1: What species are included in this dataset?
Let's inspect the unique values in the species name column to see what we're tracking.

````python
import pandas as pd

# Load the raw dataset
df = pd.read_csv("../04_data/raw/01_smroadrunners_data.csv")

# Grab a list of unique species names
unique_species = df['common_name'].unique()

print(f"Found {len(unique_species)} unique species entries:")
for species in unique_species:
    print(f" - {species}")
````

Now let's find out how many observations were collected for each species:

````python
import pandas as pd

# Load the raw dataset
df = pd.read_csv("../04_data/raw/01_smroadrunners_data.csv")

# Grab a list of unique species names
unique_species = df['common_name'].unique()

# Get unique species names and their observation counts (sorted highest to lowest)
species_counts = df['common_name'].value_counts()

print(f"Found {len(species_counts)} unique species entries:\n")
for species, count in species_counts.items():
    print(f" - {species}: {count} observations")
````

### Step 2: Sorting the Data
Pandas lets us sort our data easily using `.sort_values()`. Let's look at our observations chronologically.

````python
import pandas as pd

# Load the raw dataset
df = pd.read_csv("../04_data/raw/01_smroadrunners_data.csv")

# Sort the data by date (oldest to newest)
# Note: use df.head() to remind yourself of column names if needed!
df_sorted_date = df.sort_values(by='observed_on', ascending=True)

print("--- Oldest Observations ---")
display(df_sorted_date[['observed_on', 'common_name']].head(3)) # This specifies that we want to see the columns called 'observed_on' and 'common_name'
````

Take a close look at the observed_on column. When we checked `df.head()` in the first code cell above, we saw `observed_on` dates like `10/11/2008` (row 1, index 0). Why isn't that date showing up with the oldest observations?

This is a good example of why data cleaning is so important. Pandas is sorting the dates as a text string, which means the first digit, in this case month, is used to organize the data. We will work on fixing this in the next notebook.

We can also sort by other variables. Let's try sorting by location.

````python
# Sort the data by location (latitude and longitude)
df_sorted_location = df.sort_values(by=['latitude', 'longitude'], ascending=[True, True])

print("\n--- Sorted by Location (Lat/Long) ---")
display(df_sorted_location[['latitude', 'longitude', 'common_name']].head(3))
````

This sorts based on latitude only! That is, it puts the numbers in order, but this is not a true spatial sort. Python offers several ways to approach spatial data. 
- Pandas has a geospatial library called geopandas, which can be used with another library called matplotlib to make static maps.
- Leaflet and Folium are libraries that enable dynamic mapping (think Google Maps).

If that worked (may depend on your ArcPro version), then run the code chunk below to plot the roadrunners sample data on a map in the notebook. If it didn't work, don't worry. We will be mapping data in the next notebook.

````python
import pandas as pd
import folium
import base64
from IPython.display import IFrame, display

# Load data
df = pd.read_csv("../04_data/raw/01_smroadrunners_data.csv")

# Create map
m = folium.Map(
    location=[
        df['latitude'].mean(),
        df['longitude'].mean()
    ],
    zoom_start=11
)

# Add points
for _, row in df.iterrows():
    folium.Marker(
        location=[row['latitude'], row['longitude']]
    ).add_to(m)

# Render map in notebook
html = m.get_root().render()
encoded = base64.b64encode(html.encode("utf-8")).decode("utf-8")

display(IFrame(
    src=f"data:text/html;base64,{encoded}",
    width="100%",
    height=600
))
````
