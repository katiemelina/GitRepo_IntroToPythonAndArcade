### Debugging: The Most Universal Coding Experience
One of the most important parts of coding is figuring out why a chunk, script, or line of code isn't working. This is called **debugging**. Python is incredibly helpful in providing us with detailed error logs called **Tracebacks**. Execute the chunk below to see an example of a traceback error.

````python
# Let's debug this cell!
# Click this cell and press Shift + Enter to run it. You will see the output just below!

message = "Hello GIS World! Python is running."
print(variable)
````

Let's take a closer look at this error statement. Generally, for a traceback, the last line tells you what you need to know. In this case, we see "NameError: name 'variable' is not defined".

What does this mean? NameError tells us that python is confused about how we've named something. In this case, we are using a name, 'variable', that was not defined in the code chunk before we called it.

This particular code chunk is small, so we can quickly find where the undefined variable is and fix it. But let's imagine that instead we had a script with hundreds of lines of code. Tracking down an error there could be difficult! Fortunately, python tells us exactly where to find the error. If we look at the top of the example error message, we see "Cell In[12], line 5". A for loop repeats an action once for each item in a list. For loops are the secret weapon of Python, because the same few lines can be applied to thousands of items. Let’s loop through the list you just made:

````python
# Uses the list from the cell above (you must either run that cell or define your list again here)
for item in my_list_name:
    print(f"Item: {item}")

print("Loop finished!")
````

Look at the code chunk above:

- `for item in my_list_name:` means “for each thing in the list, do the following.” `item` is a temporary variable holding one list entry at a time. You can name it anything. The `colon` tells Python that an indented block comes next.
- The indented line (4 spaces, or press Tab) is the loop’s body. Indentation is how Python knows which lines belong inside the loop. That line runs once per `item`.
- `print("Loop finished!")` is not indented, so it is outside the loop and runs once, after the loop is done.

#### Dictionaries
A list stores items by position (e.g. [0], [1]). A dictionary stores items by *label*. Every entry in a dictionary consists of a label (also called a *key*), a colon, and a *value*. 

````python
# A dictionary stores labeled key:value pairs. Think of it as ONE ROW of a spreadsheet:
# the column name is the key, and the cell contents are the value.
# take a look at the observation dictionary below. There are five key:value pairs
observation = {
    "id": 101,
    "species": "Greater Roadrunner",
    "year": 2019,
    "latitude": 35.08,
    "longitude": -106.65
}

print(observation["species"])   # look up a value using its key
print(observation["year"])
````

There are some important syntax elements to remember with dictionaries:
- Curly brackets `{}` define a dictionary.
- Each entry in the dictionary is a key:value pair, separated with commas
- Lists find a position using an index (e.g. wildlife_species[0]). Dictionaries find a value using the key (e.g. observation["species"])
- A DataFrame is kind of like a big set of dictionaries (rows = values, columns = keys). We will learn more about data frames below!

##### If/Else
These statements allow a code chunk to make a decision using boolean values.

````python
year = 2012

if year < 2015:
    print("Older record")
else:
    print("Recent record")
````

- `if` is followed by a boolean test (True/False) and a colon.
- If the test is True, then the first indented line runs.
- `else` runs if the test is False
- Comparison symbols: `==` equal to, `!=` not equal to, `<` less than, `>` greater than, `<=` and `>=`. **A single `=` assigns a value to a variable. A double `==` evaluates whether two things are equal.**

We can add in additional tests using `elif`:

````python
for obs in observations:
    if obs["year"] < 2018:
        print(f"{obs['species']} ({obs['year']}): older record")
    elif obs["year"] == 2018:
        print(f"{obs['species']} ({obs['year']}): exactly 2018")
    else:
        print(f"{obs['species']} ({obs['year']}): recent record")
````

### Step 3: What is a Pandas DataFrame?
In Excel, you have a spreadsheet with rows and columns. The same is true for an attribute table in ArcPro. In Python, that table is called a **DataFrame**. 

For this workshop the instructor has downloaded several datasets from iNaturalist.org. Let's load 01_smroadrunners_data.csv. This is just a tiny snippet of the full roadrunner dataset, but it will give us a chance to see what a DataFrame looks like.

````python
import pandas as pd # pd is the conventional shorthand for pandas imports - this just makes it a little faster to call pandas functions when typing

# Load our sample CSV and name it df (for data frame)
df = pd.read_csv("../04_data/raw/01_smroadrunners_data.csv")

# Look at the first 3 rows using .head()
display(df.head(3))

# Check the shape (number of rows, number of columns)
print(f"Our sample dataframe has {df.shape[0]} rows and {df.shape[1]} columns.")
````

Since we are using a Jupyter notebook, you get to see a nice printed snippet of the table! 

Take a look at the print statement. To see the shape of the data frame, we use `df.shape[]`. This returns a (rows, columns) list. To pull out the number of rows and number of columns, we use the list index. In this case, the list has two items, so two indices. To get the number of rows we use `df.shape[0]` and for columns we use `df.shape[1]`. This is a quirk of python and many other coding languages, which start indexing at 0 instead of 1. 

If you get a `FileNotFoundError`: Python looked for the file at the path inside `read_csv()` and didn’t find it.
