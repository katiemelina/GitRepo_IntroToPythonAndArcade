### A note on AI
AI is, as everywhere, proliferating in the coding space. Whether you choose to use AI as a coding assistant or not is a personal choice. AI certainly has many drawbacks and issues. The notes in this section are not intended to endorse AI use or to be a guide for AI use in coding, However, given that much coding is a matter of pattern matching and customization, AI *can* be a helpful tool. **But** there are some important considerations.

#### Consideration 1: Data Security
Never give AI any code chunks with hard coded sensitive information. This might include user names, passwords, login portal locations, and data locations. It might also include actual data. When in doubt, it is better to google the base package or find a code snippet example that shows the structure you need to follow.

#### Consideration 2: Refactoring and Assumptions
When you edit your code, particularly when you try to streamline chunks or clean up extra lines, this is called refactoring. AI is very quick to refactor by making random changes to your code. Sometimes these changes are helpful. Sometimes they introduce additional errors or break connections later in the script. Always make sure you understand **why** choices were made in the code. 

This same applies to assumptions. AI tries to be helpful but sometimes this means it will skip asking questions or explaining all of the arguments that can be specified in a function. This means you may end up running code that doesn't do what you thought it would.

#### Consideration 3: Hallucinations
Code debugging is often iterative. This leads to issues with refactoring and assumptions, as discussed above. But it can also lead to long sessions, over the course of which AI can forget context and begin to hallucinate. It is good practice to keep a running log of changes to a script (e.g. through # comments or in a companion ipynb) so that you can avoid scope creep, maintain perspective, and make sure you don't lose hours of code to AI changes. Also, always dev (develop) on a copy of the code in case you need to revert to an earlier iteration. Jupyter notebooks can be very helpful for this as you can split your code into chunks, add markdown comments, and outline why you made certain changes.
